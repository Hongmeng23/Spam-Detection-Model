```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
spam = pd.read_csv("spam.csv", encoding='latin-1') #had to use the latin-1 translator because it is an old dataset and the encoding used isn't utf-8

spam = spam.rename(columns={"v1": "label", "v2": "text"})
spam = spam.drop(columns=["Unnamed: 2", "Unnamed: 3", "Unnamed: 4"])
spam["label"] = spam["label"].map({"ham": 0, "spam": 1})
spam.head()
```

```python
spam.shape
```

```python
spam.isnull().sum().sort_values(ascending=False)
```

```python
print(spam["label"].value_counts(dropna=False))
```

```python
print(spam["text"].value_counts(dropna=False))
```

```python
print("Number of Duplicate rows", spam.duplicated().sum())
spam[spam.duplicated(keep=False)].sort_values("label")
```

```python
spam = spam.drop_duplicates()
print("Number of Duplicate rows", spam.duplicated().sum())
```

```python
spam.to_csv("cleaned_spam_partial.csv", index=False)
```
