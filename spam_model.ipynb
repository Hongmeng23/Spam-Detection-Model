```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report, accuracy_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression

import pandas as pd
import matplotlib.pyplot as plt
```

```python
spam = pd.read_csv("cleaned_spam_partial.csv")
```

```python
x = spam["text"]
y = spam["label"]

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)

print("Train: ", x_train.shape, "Test: ", x_test.shape)
```

```python
spam_model = make_pipeline(TfidfVectorizer(lowercase=True, stop_words="english"), MultinomialNB())

spam_model.fit(x_train, y_train)

pred = spam_model.predict(x_test)
print("Naive bayes accuracy: ", round(accuracy_score(y_test, pred), 2))
print(classification_report(y_test, pred, target_names=["Ham", "Spam"]))
```

```python
new_message = pd.DataFrame({"text":["Congatulatons! You won a free ticket, call now !"]})
pred = spam_model.predict(new_message)[0]
proba = spam_model.predict_proba(new_message)[0]

print(f"Predicted class : {pred} (0 = ham and 1 = spam)")
print(f"Probability of messaage being Ham: {proba[0]:.2f}")
print(f"Probability of message being Spam: {proba[1]:.2f}")

#Got a wrong prediction the model was supposed to predict spam but instead predicted ham and i think this was because there was no precedent of my custom message in the training data so i tested it with a message from the data set to test the model 
```

```python
# Find an actual spam message and it still predicted ham and earlier the model recalled about 74% of the spam message and missed 26% which means this message is among the 26% missed and it could also be because of the class imbalance and the threshold. i'll be testing with a logistic regression algorithm instead to know if it'll catch this message
actual_spam_message = x_test[y_test == 1].iloc[0]
print("Actual message:", actual_spam_message)


test_df = pd.DataFrame({"text": [actual_spam_message]})
print("Prediction:", spam_model.predict(test_df)[0])
```

```python
#Testing with the logisitic regression algorithm
spam_model = make_pipeline(TfidfVectorizer(lowercase=True, stop_words="english"), LogisticRegression())

spam_model.fit(x_train, y_train)

pred = spam_model.predict(x_test)
print("Naive bayes accuracy: ", round(accuracy_score(y_test, pred), 2))
print(classification_report(y_test, pred, target_names=["Ham", "Spam"]))
```

```python
new_message = pd.DataFrame({"text":["Congatulatons! You won a free ticket, call now !"]})
pred = spam_model.predict(new_message)[0]
proba = spam_model.predict_proba(new_message)[0]

print(f"Predicted class : {pred} (0 = ham and 1 = spam)")
print(f"Probability of messaage being Ham: {proba[0]:.2f}")
print(f"Probability of message being Spam: {proba[1]:.2f}")

#Logistic regression was able to predict that this was spam unlike Naive bayes, although the recall did drop to 68% for logisitc regression 
```
